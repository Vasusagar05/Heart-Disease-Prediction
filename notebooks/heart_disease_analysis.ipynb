# Heart Disease Prediction — Exploratory Data Analysis

This notebook inspects the local UCI CSV before modeling. The original Cleveland target is `num` (severity 0–4); positive values are mapped to heart disease class 1. No model performance values are assumed here.

```python
from pathlib import Path
import sys
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
ROOT = Path.cwd()
if not (ROOT / 'data' / 'heart.csv').exists(): ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'src'))
from preprocessing import load_dataset, normalize_target
df = load_dataset(ROOT / 'data' / 'heart.csv')
df, target = normalize_target(df)
print('Target column:', target)
df.head()
```

## Dataset overview
Shape, schema, missing data, duplicates, and descriptive statistics are computed from the supplied file.

```python
print('Shape:', df.shape)
display(df.dtypes.rename('dtype').to_frame())
display(df.isna().sum().rename('missing').to_frame())
print('Duplicate records:', df.duplicated().sum())
display(df.describe(include='all').T)
```

## Distributions and target comparisons

```python
sns.set_theme(style='whitegrid')
fig, axes = plt.subplots(2, 3, figsize=(10, 9))
sns.countplot(data=df, x=target, ax=axes[0,0]); axes[0,0].set(title='Target distribution', xlabel='Heart disease (0/1)')
sns.histplot(data=df, x='age', hue=target, bins=20, kde=True, ax=axes[0,1]); axes[0,1].set(title='Age distribution')
sns.countplot(data=df, x='sex', hue=target, ax=axes[0,2]); axes[0,2].set(title='Gender by target')
sns.histplot(data=df, x='chol', hue=target, bins=20, kde=True, ax=axes[1,0]); axes[1,0].set(title='Cholesterol distribution')
sns.histplot(data=df, x='thalach', hue=target, bins=20, kde=True, ax=axes[1,1]); axes[1,1].set(title='Maximum heart rate')
sns.histplot(data=df, x='trestbps', hue=target, bins=20, kde=True, ax=axes[1,2]); axes[1,2].set(title='Resting blood pressure')
plt.tight_layout()
```

```python
numeric = df.select_dtypes(include='number')
plt.figure(figsize=(10, 9)); sns.heatmap(numeric.corr(), cmap='coolwarm', center=0, annot=True, fmt='.2f'); plt.title('Feature correlation heatmap'); plt.tight_layout()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(10, 4))
for ax, feature in zip(axes, ['age', 'chol', 'thalach']):
    sns.boxplot(data=df, x=target, y=feature, ax=ax); ax.set_title(f'{feature} by target')
plt.tight_layout()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(10, 4))
for ax, feature in zip(axes, ['cp', 'exang', 'slope']):
    sns.countplot(data=df, x=feature, hue=target, ax=ax); ax.set_title(f'Disease distribution by {feature}')
plt.tight_layout()
```

## Interpretation note
These plots describe this dataset only. They do not establish causation or clinical usefulness. Training, held-out metrics, and plots are produced reproducibly by `python src/train.py` and saved under `reports/`.
